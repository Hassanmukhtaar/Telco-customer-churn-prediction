## Logistic Regression

```python
X = pd.get_dummies(df.drop(columns="Churn"), drop_first=True)
y = df["Churn"]
```

```python
X_norm = StandardScaler().fit(X).transform(X)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X_norm, y, test_size=0.2, random_state=42)
```

```python
LR = LogisticRegression().fit(X_train,y_train)
```

```python
yhat = LR.predict(X_test)
yhat[:10]
```

```python
y_test
```

```python
print(classification_report(y_test, yhat))
print("ROC-AUC:", round(roc_auc_score(y_test, yhat), 3))
```

```python
LR.score(X_train, y_train)
```

```python
accuracy_score(y_test, yhat)
```

```python
accuracy_score(y_test, yhat)
```

```python
coefficients = pd.Series(LR.coef_[0], index=X.columns)
top = coefficients.reindex(coefficients.abs().sort_values(ascending=False).head(15).index)

plt.figure(figsize=(8, 6))
top.sort_values().plot(kind="barh", color="steelblue")
plt.title("Logistic Regression - Top 15 Feature Coefficients")
plt.xlabel("Coefficient Value (positive = more churn)")
plt.tight_layout()
plt.show()
```

## Random Classifier

```python
n_estimators=100
rf = RandomForestClassifier(n_estimators=n_estimators, random_state=42)
```

```python
rf.fit(X_train, y_train)
rf_prob = rf.predict_proba(X_test)[:, 1]
```

```python
print("Random Forest")
print(classification_report(y_test, rf.predict(X_test)))
print("ROC-AUC:", round(roc_auc_score(y_test, rf_prob), 3))
```

```python
rf.score(X_train,y_train)
```

```python
# Graph 2: Feature importance (top 15)
importances = pd.Series(rf.feature_importances_, index=X.columns).sort_values(ascending=False).head(15)

plt.figure(figsize=(8, 6))
importances.sort_values().plot(kind="barh", color="seagreen")
plt.title("Random Forest - Top 15 Feature Importances")
plt.xlabel("Importance")
plt.tight_layout()
plt.show()
```
